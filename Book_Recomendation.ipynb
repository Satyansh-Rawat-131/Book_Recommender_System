```python
num_ratings = ratings_with_books.groupby('title')['rating'].count().reset_index()
```

```python
num_ratings.rename(columns = {'rating' : 'num_of_ratings'} , inplace = True)
```

```python
num_ratings
```

```python
final_rating = ratings_with_books.merge(num_ratings , on = 'title')
```

```python
final_rating.head()
```

```python
final_rating = final_rating[final_rating['num_of_ratings']>=50]
```

```python
final_rating.shape
```

```python
final_rating.drop_duplicates(['user_id' , 'title'] , inplace = True)
```

```python
final_rating.shape
```

```python
book_pivot = final_rating.pivot_table(columns = 'user_id' , index = 'title' , values = 'rating')
```

```python
book_pivot
```

```python
book_pivot.shape # users reduced because they were not rating the books that have less than 50 reviews
```

```python
# prob = very much null value solve it
book_pivot.fillna(0 , inplace = True)
```

```python
book_pivot
```

```python
# trick
# will convert the matrix into sparse matrix because book_pivot conatains many zeros which will increase time and diffcuilty in claculating the distance in NN algo to save compute power and for simple calculation
from scipy.sparse import csr_matrix
book_sparse = csr_matrix(book_pivot)
```

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
model = NearestNeighbors(algorithm = 'brute')
```

```python
model.fit(book_sparse)
```

```python
distances , suggestions = model.kneighbors(book_pivot.iloc[237 , :].values.reshape(1 , -1) , n_neighbors = 6)# i wnat a row from book_pivot and wnats its suggestions (nearest neighbour)
```

```python
distances # distances of similar books
```

```python
suggestions # books suggested by my model
```

```python
for i in range(len(suggestions)):
  print(book_pivot.index[suggestions[i]])
```

```python
(book_pivot.index == 'The Cradle Will Fall')
```

```python
# trick
# when using np.where on an array full of boolean value it gives the index of the true value
np.where(book_pivot.index == 'The Cradle Will Fall')[0]  [0]
```

```python
# creating the functions
def recommend_book_name(book_name):
  book_id = np.where(book_pivot.index == book_name)[0]  [0]
  distances , suggestions = model.kneighbors(book_pivot.iloc[book_id , :].values.reshape(1 , -1) , n_neighbors = 6)

  for i in range(len(suggestions)):
      print(book_pivot.index[suggestions[i]])
```

```python
recommend_book_name('Animal Farm')
```
